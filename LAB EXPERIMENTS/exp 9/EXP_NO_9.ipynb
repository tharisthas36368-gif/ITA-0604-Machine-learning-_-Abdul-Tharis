```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score

rng = np.random.default_rng(1)
X = np.sort(rng.uniform(-3, 3, 120)).reshape(-1, 1)
y = 0.5 * X.ravel() ** 3 - X.ravel() ** 2 + 2 * X.ravel() + rng.normal(0, 1.5, 120)

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=0)

lin = LinearRegression().fit(X_train, y_train)
lin_pred = lin.predict(X_test)

results = {"Linear": (r2_score(y_test, lin_pred), mean_squared_error(y_test, lin_pred))}
models = {"Linear": (lin, None)}
for degree in [2, 3, 4]:
    poly = PolynomialFeatures(degree=degree)
    model = LinearRegression().fit(poly.fit_transform(X_train), y_train)
    pred = model.predict(poly.transform(X_test))
    results[f"Polynomial (degree {degree})"] = (r2_score(y_test, pred),
                                                mean_squared_error(y_test, pred))
    models[f"Polynomial (degree {degree})"] = (model, poly)

print(f"{'Model':<24}{'R2 score':>10}{'MSE':>10}")
for name, (r2, mse) in results.items():
    print(f"{name:<24}{r2:>10.4f}{mse:>10.4f}")

plt.scatter(X, y, color="gray", s=15, label="Data")
for name, (model, poly) in models.items():
    Xp = X if poly is None else poly.transform(X)
    plt.plot(X, model.predict(Xp), label=name)
plt.xlabel("X")
plt.ylabel("y")
plt.title("Linear vs Polynomial Regression")
plt.legend()
plt.show()
```
