```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

rng = np.random.default_rng(0)
experience = rng.uniform(0, 15, 100)
salary = 30000 + 5000 * experience + rng.normal(0, 4000, 100)
pd.DataFrame({"YearsExperience": experience, "Salary": salary}).to_csv("salary.csv", index=False)

df = pd.read_csv("salary.csv")
X = df[["YearsExperience"]]
y = df["Salary"]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)

model = LinearRegression()
model.fit(X_train, y_train)
y_pred = model.predict(X_test)

print("Intercept :", round(model.intercept_, 2))
print("Slope     :", round(model.coef_[0], 2))
print("MAE       :", round(mean_absolute_error(y_test, y_pred), 2))
print("MSE       :", round(mean_squared_error(y_test, y_pred), 2))
print("RMSE      :", round(np.sqrt(mean_squared_error(y_test, y_pred)), 2))
print("R2 score  :", round(r2_score(y_test, y_pred), 4))

new = pd.DataFrame({"YearsExperience": [7.5]})
print("Predicted salary for 7.5 years of experience:", round(model.predict(new)[0], 2))

plt.scatter(X_train, y_train, color="blue", label="Training data")
plt.scatter(X_test, y_test, color="green", label="Test data")
plt.plot(X, model.predict(X), color="red", label="Regression line")
plt.xlabel("Years of Experience")
plt.ylabel("Salary")
plt.title("Linear Regression")
plt.legend()
plt.show()
```
