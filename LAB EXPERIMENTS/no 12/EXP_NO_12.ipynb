```python
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import (accuracy_score, confusion_matrix,
                             ConfusionMatrixDisplay, classification_report)

iris = load_iris()
X_train, X_test, y_train, y_test = train_test_split(
    iris.data, iris.target, test_size=0.2, random_state=42, stratify=iris.target)

scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

accuracies = []
k_values = range(1, 21)
for k in k_values:
    accuracies.append(accuracy_score(
        y_test, KNeighborsClassifier(n_neighbors=k).fit(X_train, y_train).predict(X_test)))
best_k = k_values[accuracies.index(max(accuracies))]
print("Best K:", best_k, "with accuracy", round(max(accuracies), 4))

model = KNeighborsClassifier(n_neighbors=5).fit(X_train, y_train)
y_pred = model.predict(X_test)
print("\nAccuracy (K=5):", round(accuracy_score(y_test, y_pred), 4))
print("Confusion matrix:\n", confusion_matrix(y_test, y_pred))
print("\nClassification report:\n",
      classification_report(y_test, y_pred, target_names=iris.target_names))

sample = scaler.transform([[5.1, 3.5, 1.4, 0.2]])
print("Predicted species for [5.1, 3.5, 1.4, 0.2]:", iris.target_names[model.predict(sample)[0]])

plt.plot(list(k_values), accuracies, marker="o")
plt.xlabel("K")
plt.ylabel("Accuracy")
plt.title("Iris - KNN accuracy for different K")
plt.show()
ConfusionMatrixDisplay(confusion_matrix(y_test, y_pred),
                       display_labels=iris.target_names).plot(cmap="Blues")
plt.show()
```
